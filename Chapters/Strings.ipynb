# Strings

## What is a string?
A string is a data type used to store text. In C#, a string holds a sequence of characters such as letters, numbers, spaces, and symbols, and it is written using double quotes. When you store a string in a variable, the variable does not hold the text itself but instead holds a reference to where that text is stored in memory. Strings are commonly used for things like names, messages, user input, and file data, and once a string is created, it cannot be changed.

```csharp
string message = "Hello, ETSU";

Console.WriteLine(message);
//The above statement prints to the console ----> Hello, ETSU

Strings are used for:

- User input
- Displaying output
- File input and output
- Storing names, sentences, and messages.
- Strings like other the other primitive data types are popular and standard in other programming languges
- Strings are a common data type to interact between two different programs

### Strings Are Objects

In C#, strings are objects, not primitive data types.

This means:
- Strings have methods
- Strings have properties
- You can call methods directly on a string variable

## What Is Actually Stored in a String Variable?

When you write:

```csharp
string word = "ETSU";
```

The variable `word` does not store the characters E T S U directly.

Instead, it stores a reference (memory location) that points to where the string data lives in memory.

Think of it this way:
- The string characters live somewhere in memory
- The variable stores the address of that location

In memory it looks like this:

**In Memory:**

`word ───▶ Memory Address 0x1234`

***The variable `word` holds where the string is, not the string itself.***

### Why does this matter?

**Example 1: Assigning One String Variable to Another**

```csharp
string a = "Hello";

string b = a;
```

**In Memory:**

`a ───▶ Memory Address 0x2000`

`b ───▶ Memory Address 0x2000`

**At Memory Address:**

`0x2000 → "Hello"` *{Collection of Character Objects}*

## Strings are Immutable

Strings cannot be changed after creation.

```csharp
string greeting = "hello";
greeting = greeting.ToUpper();

Console.WriteLine(greeting);
// Output: HELLO

What actually happens:
1. `"hello"` is created in memory
2. `"HELLO"` is created in a new location using `word.ToUpper()`
3. `word` is updated to point to the new location

Before:

`word ───▶ "hello"`

After:

`word ───▶ "HELLO"`

*The original "hello" still exists temporarily until the Garbage Collector removes it.*

## Comparing Strings 

### Value Comparison

Below, C# compares string contents, not memory locations.

```csharp
string x = "ETSU";
string y = "ETSU";

Console.WriteLine(x == y); // true
//Output: true

### Reference Comparison

Below checks if both variables point to the same memory location.

Sometimes this returns true, sometimes false, because of string interning.

```csharp
string x = "ETSU";
string y = "ETSU";

Console.WriteLine(object.ReferenceEquals(x, y)); //True because string interning realizes its the same value

string a = "ET";
string b = "SU";

string c = $"{a}{b}";

Console.WriteLine(object.ReferenceEquals(x, c)); //False, c is a new memory location of the combined letter

Console.WriteLine(x == c); //True because the values are the same

### String Interning (Important but Simple)

C# keeps a pool of common strings.

If you write:

```csharp
string a = "CSCI";

string b = "CSCI";
```

C# may store `CSCI` once and point both variables to it.

This improves memory efficiency.

## Escape Characters

Sometimes you need to include special characters inside a string, such as a new line, a tab, or quotation marks. In C#, these are written using escape characters. An escape character starts with a backslash \ and tells the program to treat the next character in a special way.

| Escape Character | Name             | Description                                   | Example Output            |
|------------------|------------------|-----------------------------------------------|---------------------------|
| `\n`             | New Line         | Moves text to the next line                   | Line1<br>Line2            |
| `\t`             | Tab              | Inserts a horizontal tab                      | Name:    Value            |
| `\"`             | Double Quote     | Displays a double quotation mark              | "Hello"                   |
| `\\`             | Backslash        | Displays a backslash                          | C:\Folder\File            |
| `\r`             | Carriage Return  | Moves cursor to start of line                 | Overwrites line start     |
| `\b`             | Backspace        | Removes previous character                    | Text correction           |
| `\a`             | Alert (Bell)     | Triggers system alert sound (if supported)   | *Beep*                    |
| `\0`             | Null             | Represents the null character                 | End of string marker      |

### Below is an example of each of the escape characters in order except for `\a` and `\0`

```csharp
Console.WriteLine("Welcome to CSCI 1250\nEast Tennessee State University");
//Output: 
//Welcome to CSCI 1250
//East Tennessee State University 

```csharp
Console.WriteLine("Name:\tMathew");
Console.WriteLine("Course:\tCSCI 1250");
Console.WriteLine("School:\tETSU");
//Output:
//Name:   Mathew
//Course: CSCI 1250
//School: ETSU

```csharp
Console.WriteLine("The instructor said, \"Read the instructions carefully.\"");
//Output: The instructor said, "Read the instructions carefully."

```csharp
Console.WriteLine("File saved to: C:\\Users\\Student\\Documents");
//Output: File saved to: C:\Users\Student\Documents

```csharp
Console.WriteLine("Loading...\rDone!");
//Output:
//Loading...
//Done!

```csharp
Console.WriteLine("Hello\b!");
//Output: Hell!

```csharp
Console.WriteLine("Item\tPrice\nApple\t$1.00\nOrange\t$0.75");
//Output:
//Item    Price
//Apple   $1.00
//Orange  $0.75
